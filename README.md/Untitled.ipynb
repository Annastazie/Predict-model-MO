# Predict-model-MO

## Описание проекта

`Predict-model-MO` — это MVP интеллектуальной системы прогнозирования итогового балла муниципального образования на основе агрегированных квартальных данных по 8 сферам деятельности.

## Цель проекта

Цель проекта — разработать прототип системы, позволяющей:

- автоматически использовать квартальные данные по муниципальным образованиям;
- прогнозировать итоговый балл муниципального образования;
- сравнивать фактическое и прогнозное значение;
- визуализировать результаты в удобном пользовательском интерфейсе.

---

## Бизнес-контекст

Проект ориентирован на задачу оценки эффективности муниципальных образований в системе государственного управления.  
В исходном процессе итоговый балл формируется по квартальным данным, агрегированным по нескольким сферам. Основная проблема заключается в том, что итоговая оценка становится доступной только после завершения периода и обработки данных.

Разработанный MVP позволяет перейти от чисто ретроспективного анализа к прогнозной аналитике и использовать исторические данные для предварительной оценки результата.

---

## Используемые данные

В проекте используются обезличенные данные за **3 квартала 2024 года**.

Каждое наблюдение соответствует одному муниципальному образованию в конкретном квартале.

### Входные признаки:
- агрегированные баллы по 8 сферам:
  - `sphere_1`
  - `sphere_2`
  - `sphere_3`
  - `sphere_4`
  - `sphere_5`
  - `sphere_6`
  - `sphere_7`
  - `sphere_8`

### Целевая переменная:
- `total_score` — итоговый балл муниципального образования

### Дополнительные признаки (feature engineering):
- `quarter_num` — числовой код квартала
- `avg_sphere_score` — средний балл по сферам
- `min_sphere_score` — минимальный балл по сферам
- `max_sphere_score` — максимальный балл по сферам
- `weak_spheres_count` — количество проблемных сфер

## Реализованные этапы проекта

В рамках проекта были реализованы следующие этапы:

1. загрузка и объединение квартальных Excel-данных;
2. обезличивание муниципальных образований;
3. формирование единого датасета;
4. проверка качества данных;
5. exploratory data analysis (EDA);
6. feature engineering;
7. хронологическое разделение train/test;
8. baseline-модель;
9. обучение и сравнение нескольких моделей;
10. анализ прогнозов и ошибок;
11. сохранение моделей и результатов экспериментов;
12. разработка демонстрационного интерфейса на Streamlit.

---

## Использованные модели

В проекте были протестированы следующие модели:

- **Baseline_mean** — простой прогноз среднего значения;
- **LinearRegression_basic** — линейная регрессия на базовом наборе признаков;
- **LinearRegression_extended** — линейная регрессия на расширенном наборе признаков;
- **Ridge_extended** — Ridge-регрессия;
- **RandomForest_extended** — случайный лес на расширенном наборе признаков.

---

## Результаты моделирования

По результатам экспериментов лучшей моделью стала:

- **RandomForest_extended**

Основные метрики лучшей модели на тестовой выборке:

- **MAE**: 0.369
- **RMSE**: 0.543
- **MAPE**: 1.77%

При этом интерпретируемая модель:

- **LinearRegression_basic**

также показала высокое качество:

- **MAE**: 0.760
- **RMSE**: 0.924
- **MAPE**: 3.38%

Таким образом, в проекте выделяются две значимые модели:

- **LinearRegression_basic** — как базовая интерпретируемая модель;
- **RandomForest_extended** — как лучшая модель по точности.

## Интерфейс

В рамках проекта реализован демонстрационный интерфейс на **Streamlit**.

Интерфейс позволяет:

- выбрать квартал;
- выбрать муниципальное образование;
- выбрать модель прогнозирования;
- увидеть фактический итоговый балл;
- получить прогноз модели;
- оценить абсолютную ошибку;
- просмотреть значения по сферам;
- просмотреть дополнительные признаки.
